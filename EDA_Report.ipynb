2. Hypothesis Testing Results
a. Hypothesis 1: Revenue and Operational Costs
Null Hypothesis (H₀): There is no significant correlation between revenue and operational costs (e.g., fuel and maintenance).

Alternative Hypothesis (H₁): There is a significant correlation between revenue and operational costs.

Test: Use Pearson's correlation to test the relationship between revenue and operational costs (fuel, maintenance).

```python
# Pearson correlation between Revenue and Fuel Costs
correlation, p_value = pearsonr(df_revenue['revenue'], df_costs['fuel'])
print(f"Pearson correlation: {correlation}, P-value: {p_value}")
```

```python
Interpretation:

If the p-value is less than 0.05, reject the null hypothesis (indicating a significant correlation).

A high correlation coefficient (close to 1 or -1) indicates a strong relationship.

b. Hypothesis 2: Customer Satisfaction vs Revenue
Null Hypothesis (H₀): Customer satisfaction does not significantly affect revenue.

Alternative Hypothesis (H₁): Customer satisfaction significantly affects revenue.

Test: Use Spearman’s rank correlation or Kendall’s Tau if the relationship is non-linear or ordinal.
```

```python
# Spearman's rank correlation between Revenue and Customer Ratings
spearman_corr, p_value = spearmanr(df_revenue['revenue'], df_ratings['rating'])
print(f"Spearman correlation: {spearman_corr}, P-value: {p_value}")
```

Interpretation:

If the p-value is less than 0.05, reject the null hypothesis.

A strong correlation coefficient indicates that customer satisfaction has a significant impact on revenue.
